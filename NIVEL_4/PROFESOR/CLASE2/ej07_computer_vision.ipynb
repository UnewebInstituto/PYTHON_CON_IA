```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive

# 1. Cargar imagen y convertir a escala de grises
drive.mount('/content/drive')

img = cv2.imread('/content/drive/MyDrive/data/imagen_arroz.png')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Binarización (Segmentación)
# Convertimos píxeles claros (arroz) a blanco y resto a negro
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# 3. Detección de contornos (Objeto)
contornos, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# 4. Dibujar resultados
resultado = img.copy()
cv2.drawContours(resultado, contornos, -1, (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(resultado, cv2.COLOR_BGR2RGB))
plt.title(f'Granos detectados: {len(contornos)}')
plt.show()
```
