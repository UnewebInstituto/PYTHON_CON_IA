```python
import tensorflow as tf

x = tf.Variable(1.0)

def f(x):
  y = x**2 + 2*x - 5
  return y

print(f'El resultado de evaluar la función es: {f(x)}')
```

```python
with tf.GradientTape() as tape:
  y = f(x)

g_x = tape.gradient(y, x)  # g(x) = dy/dx

print(f'El resultado de evaluar la 1ra. derivada de la función es: {g_x}')
```
