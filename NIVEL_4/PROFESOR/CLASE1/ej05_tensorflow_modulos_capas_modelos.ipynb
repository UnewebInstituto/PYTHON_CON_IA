```python
import tensorflow as tf

# Declaración de la clase
class MyModule(tf.Module):
  def __init__(self, value):
    self.weight = tf.Variable(value)

  @tf.function
  def multiply(self, x):
    return x * self.weight
```

```python
mod = MyModule(3)
mod.multiply(tf.constant([1, 2, 3]))
```
