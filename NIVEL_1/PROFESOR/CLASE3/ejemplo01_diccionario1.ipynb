```python
# Ejemplo 1 Manejo de Diccionario
data = ["A","B", "A","C"]
conteo = {}
for x in data: conteo[x] = conteo.get(x,0) + 1
print(f"Frecuencia de categorías:{conteo}")
```

```python
# Ejemplo 2 Manejo de Diccionario
pedido = {"prod1":{"precio":10,"cantidad":2}, "prod2":{"precio":5,"cantidad":3}}
total = sum(v["precio"] * v["cantidad"] for v in pedido.values())
print(f"El total del pedido es:{total}")
```

```python
# Ejemplo 3 Manejo de Diccionario
stats = {"clase_A":{"media":50,"std":5}, "clase_B":{"media":60,"std":8}}
print(f"Media de la clase A:{stats["clase_A"]["media"]}")
```

```python
# Ejemplo 4 Manejo de Diccionario
config = {"tasa":0.01,"tiempo":60}
config["tasa"] *= 100
print(f"Nueva configuración de tasa:{config["tasa"]}")
```
