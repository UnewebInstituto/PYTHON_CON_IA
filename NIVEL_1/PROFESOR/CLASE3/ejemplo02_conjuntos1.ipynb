```python
# Ejemplo 1 de Conjuntos
ids = [101, 102, 101, 103]
unicos = len(set(ids))
print(f"Total de Usuarios únicos detectados: {unicos}")
```

```python
# Ejemplo 2 de Conjuntos
grupo1 = {101, 102}; grupo2 = {102, 103}
# Mostrar la intercepción entre ambos conjuntos
print(f"Ids, presentes en ambos grupos {grupo1.intersection(grupo2)}")
```

```python
# Ejemplo 3 de Conjuntos
vistos_ayer = {1,2,3} ; vistos_hoy = {3,4,5}
# Mostrar la diferencia entre ambos conjuntos
print(f"Nuevos usuarios deectados hoy : {vistos_hoy.difference(vistos_ayer)}")
```
