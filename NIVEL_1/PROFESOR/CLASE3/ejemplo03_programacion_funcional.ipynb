```python
# Ejemplo 1 programación funcional
precios = [100, 250, 380, 400, 560]
precios_con_iva = [p * 1.16 for p in precios]
print(f"Precios finales : {precios_con_iva}")
```

```python
# Ejemplo 2 función lambda
cuadrado = lambda x : x**2
print(f"El cuadrado de 7 es: {cuadrado(7)}")
```

```python
# Ejemplo 3 función map()
str_nums = ["10","20","30","40"]
ints = list(map(int, str_nums))
print(f"Lista de enteros {ints}")
```

```python
# Ejemplo 4 función filter()
lecturas = [22, 28, 19, 31, 34]
lecturas_criticas = list(filter(lambda x : x > 25, lecturas))
print(f"Lecturas que exceden el límite : {lecturas_criticas}")
```

```python
# Ejemplo 5 Pipeline (entubado de todas las operaciones)
datos = [19, -5, 20, -1, 30]
# Filtra positivos, aplica cuadrado, convierte a lista
procesados = list(map(lambda x : x**2, filter(lambda x : x > 0 , datos)))
print(f"Datos procesados estadisticamente:{procesados}")
```
